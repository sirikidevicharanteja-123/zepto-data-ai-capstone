```python
#import required libraries
import pandas as pd
```

```python
#loading the dataset
df=pd.read_csv("data\\raw_data.csv")
#displaying The first five rows in the datset
df.head()
```

```python
#Displaying information about the dataset
df.info()
```

```python
#ch3cking for null values in the dataset
df.isnull().sum()
```

***Cleaning Price column***

```python
#removing £ in price column and converting it to float in new column price_gbp
df['price_gbp']=df['price'].str.replace('£','').str.strip().astype(float)
```

```python
df.head()
```

```python
#checking values in star_rating column
df['star_rating'].unique()
```

```python
#Converting star_ratings text into integers(1-5)
text_ratings={
  "One":1,
  "Two":2,
  "Three":3,
  "Four":4,
  "Five":5
}
df['rating']=df['star_rating'].replace(text_ratings).astype(int)
```

```python
# Create a boolean column:
# True  -> Book is in stock
# False -> Book is not in stock
df["in_stock"] = df["availability"].str.contains(
    "In stock",
    case=False,
    na=False
)
```

```python
#converting Gbp to Inr in new column price_inr
gbp_to_inr=105.50
df['price_inr']=(df['price_gbp']*gbp_to_inr).round(2)
```

```python
df.head()
```

```python
df.info()
```

```python
#Dropping the columns which are not required for analysis
df.drop(columns=["price","star_rating","availability"],inplace=True)
```

```python
#save cleaned Dataset
df.to_csv("data\\cleaned_data.csv",index=False)
print("Dataset saved Successfully")
```
